```python
import os
import requests
from tqdm import tqdm

# ✅ Update this path to where your .txt files are stored
TEXT_FILE_DIR = "/mnt/object/LoanData/eval/eval_text_files"  # or wherever your text files are

# ✅ Update this to your FastAPI server endpoint
FASTAPI_URL = "http://fastapi_server:8000/predict_loan_risk"

# List all .txt files in the directory
file_paths = [os.path.join(TEXT_FILE_DIR, f) for f in os.listdir(TEXT_FILE_DIR) if f.endswith(".txt")]

# Submit each file
for path in tqdm(file_paths, desc="Sending files"):
    with open(path, "rb") as f:
        files = {"file": (os.path.basename(path), f, "text/plain")}
        try:
            res = requests.post(FASTAPI_URL, files=files, timeout=5)
            if res.status_code != 200:
                print(f"Error for {path}: {res.status_code}, {res.text}")
        except Exception as e:
            print(f"Failed request for {path}: {e}")
```
